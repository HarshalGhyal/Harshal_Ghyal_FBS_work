#### Bar plot
1.Comparing categories

2.Showing frequencies and count


```python
import matplotlib.pyplot as plt
```

```python
products=['Hot Cofee','Cold Cofee','Tea','Salad']
summer=[30,90,80,50]
```

```python
plt.figure(figsize=(7,3))  # width,height
plt.bar(products,summer,color='purple')
plt.title('sales of product for summer')
plt.xlabel('products')
plt.ylabel('count')
plt.show()
```
